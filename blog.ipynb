# What Makes a Taylor Swift Song Successful?

Taylor Swift is one of the most successful music artists of all time. She has sold over 100 million album units, and multiple of her songs have surpassed one billion streams. But what actually drives her success? Is there something in the music itself that explains her dominance, or is it something else entirely?

This blog analyses the musical features of Taylor Swift’s music to see whether certain characteristics are associated with success. Are there patterns in which tracks become globally popular while others remain less prominent? Or is her success a product of more than just the music itself?

## Where does the data come from?

Data for this project came from a range of sources. Album-level success data was scraped from Wikipedia, while track-level data was collected from a Kaggle dataset and APIs from Spotify and Last.fm.

Taylor Swift was an ideal artist to study because there is an unusually large amount of data available on her individual tracks. Features such as tempo, key and energy are available at song level, allowing for a much more detailed analysis than would be possible for many other artists.

## How is success measured?

Success is measured in different ways depending on the level of analysis.

- **Album success** is measured using total certified units sold  
- **Track success** is measured using listener counts and play count  

This allows both a broad view of album performance and a more detailed track-level analysis.

The most successful album by a large margin is *1989*, with over 20 million units sold in total. Around 15 million came from its original release, with a further 5 million from the *Taylor’s Version* re-release.

More broadly, album success appears remarkably consistent across Swift’s career. While older albums do have higher total unit counts, these have also had longer to accumulate sales, so this should not be interpreted as clear evidence that earlier music was inherently stronger.

Looking at individual tracks, a small number of songs dominate total listeners while the majority receive far fewer. This creates a clear “long tail” distribution, where only a few songs become extremely large hits.

Among the very top songs, *1989* is especially strong, but there is still a relatively balanced spread across albums overall. Again, older songs have had longer to build up listens, so age is likely part of this pattern.

<h2 id="time">How has Taylor Swift’s music changed over time?</h2>

Taylor Swift is often described as a “chameleon” because of her ability to shift both genre and sound. Her earliest albums, such as *Taylor Swift* and *Fearless*, are clearly country, while later albums like *1989* and *Red* move much further toward pop. More recent albums explore indie, folk-pop and alternative influences.

However, when looking at measurable track features, her music is often more stable over time than this reputation might suggest.

This suggests that although her overall musical profile remains fairly stable, some stylistic elements do shift in meaningful ways.

This is an important result. It suggests that Taylor Swift’s popularity is not easily explained by measurable audio features such as energy, tempo or danceability alone.

## Are some features too similar to each other?

<div class="center-figure">
  <img src="output/figures/correlation_matrices/track_feature_correlations.png" alt="Correlation matrix of track features and success">
</div>

The correlation matrix shows that some features are strongly related to each other. In particular, energy, loudness and valence are all quite closely associated.

This matters because highly correlated predictors can make regression coefficients less stable and harder to interpret. In other words, if several variables are capturing very similar information, it becomes harder for the model to separate their individual effects.

<h2 id="regression">What do the regression results show?</h2>

To test these relationships more formally, I estimated regressions using track-level musical features to predict success, measured as log listeners.

The results were consistent with the visual analysis:

- coefficients were generally small  
- explanatory power was low  
- and no simple combination of musical features strongly predicted which songs would become especially popular  

This suggests that standard audio features are only a weak explanation of success within Taylor Swift’s discography.

Taken together, these results suggest that Taylor Swift’s success is not primarily determined by observable musical features.

This is perhaps the most interesting finding of the project. Despite analysing a wide range of audio characteristics, there is little evidence that any single one consistently explains why some songs become much more popular than others.

Instead, success likely depends on broader factors beyond the music itself, such as cultural influence, fan engagement, branding, timing, and narrative.

## Conclusion

This project set out to test whether musical features could explain the success of Taylor Swift’s songs.

While her music does evolve across eras, especially in features such as acousticness and explicitness, the broader conclusion is that measurable audio features alone do not strongly predict popularity. The relationships are weak, the regressions have low explanatory power, and much of Swift’s success appears to come from factors beyond the sound of the music itself.

In that sense, Taylor Swift’s success seems to be about more than just the songs. It reflects a wider combination of identity, cultural relevance and audience connection that cannot be fully captured through track-level features alone.
